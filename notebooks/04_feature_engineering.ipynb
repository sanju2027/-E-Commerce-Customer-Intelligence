```python
# Create RFM scores

# — Recency	recency_days	How recently the customer purchased
# F — Frequency	total_orders	How many orders the customer made
# M — Monetary	total_spent	How much the customer spen

```

```python
# Create RFM scores
#We'll use 5 quantile groups (1–5) so customers are ranked relative to other customers.#

#Create F_score
customer_360["R_score"] = pd.qcut(
    customer_360["recency_days"],
    q = 5,
    labels = [5,4,3,2,1],
    duplicates= "drop"
)
```

```python
customer_360["F_score"] = pd.qcut(
    customer_360["total_orders"].rank(method = "first"),
    q=5,
    labels = [1,2,3,4,5],
    duplicates = "drop"
)
```

```python
# Create M_score
# Higher spending = better:

customer_360["M_score"] = pd.qcut(
    customer_360["total_spent"],
    q = 5,
    labels = [1,2,3,4,5],
    duplicates= "drop"
)
```

```python
# check the scores
customer_360[
    ["recency_days", "total_orders", "total_spent",
     "R_score", "F_score", "M_score"]  
].head()
```

```python
customer_360[
    ["R_score", "F_score", "M_score"]
].describe()
```

```python
# Create the combined RFM_score

customer_360["RFM_score"] = (
    customer_360["R_score"].astype(int)
    + customer_360["F_score"].astype(int)
    + customer_360["M_score"].astype(int)
)
```

```python
# Create customer segments
# RFM Score	Segment
# 13–15	High Value
# 10–12	Loyal
# 7–9	Potential
# 3–6	Low ValueR
```

```python
customer_360["customer_segment"] = pd.cut(
    customer_360["RFM_score"],
    bins = [2,6,9,12,15],
    labels = [
        "Low Value",
        "Potential",
        "Loyal",
        "High Value"
    ]
)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

scaled_values = scaler.fit_transform(
    customer_360[
        ["total_spent", "total_orders"]
    ]
)

customer_360["spend_score"] = scaled_values[:, 0]
customer_360["frequency_score"] = scaled_values[:,1]
```

```python
customer_360["customer_value_score"] = (
    0.7 * customer_360["spend_score"] +
    0.3 * customer_360["frequency_score"]
)
```
