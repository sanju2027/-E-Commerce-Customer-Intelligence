```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders[col] = pd.to_datetime(orders[col], errors = "coerce")
```

```python
# Check the date range

print("Earliest order:",
      orders["order_purchase_timestamp"].min())

print("Latest order:",orders["order_purchase_timestamp"].max()

)
```

```python
# Create Delivery Days

orders["delivery_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_purchase_timestamp"]
).dt.total_seconds() / (24 * 60 *60)
```

```python
orders["delivery_days"].head()
```

```python
orders["delivery_days"].max()
```

```python
orders["delivery_days"].min()
```

```python
# Estimated Delivery days
```

```python
orders["estimated_delivery_days"] = (
   orders["order_estimated_delivery_date"]
   - orders["order_purchase_timestamp"] 
).dt.total_seconds() / (24 * 60 * 60)
```

```python
orders["estimated_delivery_days"]
```

```python
# Calculate Delivery Delay

orders["delivery_delay_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_estimated_delivery_date"]
).dt.total_seconds() / (24 * 60 * 60)
```

```python
orders["delivery_delay_days"]
```

```python
# Delivery Classification

orders["delivery_performance"] = np.where(
   orders["order_delivered_customer_date"].isna(),
   "Not Delivered",
   np.where(
       orders["delivery_delay_days"] > 0,
    "Late",
    "On Time / Early"
)
)
```
